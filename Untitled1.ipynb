```python
# ============================================================
# 3. IMPORTS
# ============================================================

import os
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, Subset

import torchvision.transforms as transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

import matplotlib.pyplot as plt
import seaborn as sns

import timm

# ============================================================
# DEVICE
# ============================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
# ============================================================
# 4. DATASET CLASS
# ============================================================

class ACRIMADataset(Dataset):

    def __init__(self, root_dir, transform=None):

        self.root_dir = root_dir
        self.transform = transform

        self.image_paths = []
        self.labels = []

        valid_extensions = (
            ".jpg",
            ".jpeg",
            ".png",
            ".bmp",
            ".tif",
            ".tiff"
        )

        for filename in sorted(os.listdir(root_dir)):

            if filename.lower().endswith(valid_extensions):

                image_path = os.path.join(
                    root_dir,
                    filename
                )

                # ------------------------------------------
                # LABEL MAPPING
                # 0 = NORMAL
                # 1 = GLAUCOMA
                # ------------------------------------------

                if "_g_" in filename.lower():
                    label = 1

                else:
                    label = 0

                self.image_paths.append(image_path)
                self.labels.append(label)

        print("Dataset loaded successfully")
        print("Total images:", len(self.image_paths))
        print("Normal:", self.labels.count(0))
        print("Glaucoma:", self.labels.count(1))

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):

        image_path = self.image_paths[idx]

        image = Image.open(image_path).convert("RGB")

        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label
# ============================================================
# 5. TRANSFORMS
# ============================================================

transform = transforms.Compose([

    transforms.Resize((224, 224)),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Transform loaded successfully!")
# ============================================================
# 6. LOAD DATASET
# ============================================================

dataset = ACRIMADataset(
    dataset_path,
    transform=transform
)

print("\nDataset size:", len(dataset))
# ============================================================
# 7. TRAIN / VALIDATION SPLIT
# ============================================================

indices = np.arange(len(dataset))

labels = np.array(dataset.labels)

train_indices, val_indices = train_test_split(
    indices,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

train_dataset = Subset(
    dataset,
    train_indices
)

val_dataset = Subset(
    dataset,
    val_indices
)

print("Total:", len(indices))
print("Training:", len(train_dataset))
print("Validation:", len(val_dataset))
# ============================================================
# 8. DATALOADERS
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
# ============================================================
# 9. CONVNEXT-TINY + MLFF
# ============================================================

class ConvNeXtMLFF(nn.Module):

    def __init__(self, num_classes=2):

        super().__init__()

        # ConvNeXt-Tiny backbone
        self.backbone = timm.create_model(
            "convnext_tiny",
            pretrained=True,
            features_only=True,
            out_indices=(1, 2, 3)
        )

        channels = self.backbone.feature_info.channels()

        print("ConvNeXt feature channels:", channels)

        # Multi-level feature projection
        self.conv1 = nn.Conv2d(
            channels[0],
            128,
            kernel_size=1
        )

        self.conv2 = nn.Conv2d(
            channels[1],
            128,
            kernel_size=1
        )

        self.conv3 = nn.Conv2d(
            channels[2],
            128,
            kernel_size=1
        )

        # MLFF
        self.fusion = nn.Sequential(

            nn.Conv2d(
                128 * 3,
                256,
                kernel_size=1
            ),

            nn.BatchNorm2d(256),

            nn.ReLU(inplace=True)
        )

        # Classifier
        self.classifier = nn.Sequential(

            nn.Dropout(0.5),

            nn.Linear(
                256,
                num_classes
            )
        )

    def forward(self, x):

        # Extract multi-level features
        features = self.backbone(x)

        f1 = self.conv1(features[0])
        f2 = self.conv2(features[1])
        f3 = self.conv3(features[2])

        # Deepest feature resolution
        target_size = f3.shape[2:]

        # Resize
        f1 = F.interpolate(
            f1,
            size=target_size,
            mode="bilinear",
            align_corners=False
        )

        f2 = F.interpolate(
            f2,
            size=target_size,
            mode="bilinear",
            align_corners=False
        )

        # Multi-Level Feature Fusion
        fused = torch.cat(
            [f1, f2, f3],
            dim=1
        )

        fused = self.fusion(fused)

        # Global average pooling
        fused = F.adaptive_avg_pool2d(
            fused,
            1
        )

        fused = torch.flatten(
            fused,
            1
        )

        # Classification
        output = self.classifier(fused)

        return output
# ============================================================
# 10. CREATE MODEL
# ============================================================

model = ConvNeXtMLFF(
    num_classes=2
).to(device)

print("\nModel created successfully!")
print("Device:", device)
# ============================================================
# 11. TRAINING SETUP
# ============================================================

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.0001,
    weight_decay=0.0001
)

epochs = 20

patience = 3

best_val_loss = float("inf")

patience_count = 0

# Google Drive model path
best_model_path = (
    "/content/drive/MyDrive/"
    "best_convnext_tiny_mlff.pth"
)

print("Training setup ready.")
# ============================================================
# 12. TRAINING
# ============================================================

for epoch in range(epochs):

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model.train()

    train_loss = 0

    train_correct = 0

    train_total = 0

    for images, labels_batch in train_loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels_batch = labels_batch.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(
            outputs,
            labels_batch
        )

        loss.backward()

        optimizer.step()

        train_loss += loss.item()

        predictions = torch.argmax(
            outputs,
            dim=1
        )

        train_total += labels_batch.size(0)

        train_correct += (
            predictions == labels_batch
        ).sum().item()

    train_loss /= len(train_loader)

    train_accuracy = (
        100 *
        train_correct /
        train_total
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model.eval()

    val_loss = 0

    val_correct = 0

    val_total = 0

    with torch.no_grad():

        for images, labels_batch in val_loader:

            images = images.to(
                device,
                non_blocking=True
            )

            labels_batch = labels_batch.to(
                device,
                non_blocking=True
            )

            outputs = model(images)

            loss = criterion(
                outputs,
                labels_batch
            )

            val_loss += loss.item()

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            val_total += labels_batch.size(0)

            val_correct += (
                predictions == labels_batch
            ).sum().item()

    val_loss /= len(val_loader)

    val_accuracy = (
        100 *
        val_correct /
        val_total
    )

    # --------------------------------------------------------
    # PRINT RESULTS
    # --------------------------------------------------------

    print(
        f"Epoch [{epoch+1}/{epochs}] "
        f"Train Loss: {train_loss:.4f} "
        f"Train Acc: {train_accuracy:.2f}% "
        f"Val Loss: {val_loss:.4f} "
        f"Val Acc: {val_accuracy:.2f}%"
    )

    # --------------------------------------------------------
    # SAVE BEST MODEL
    # --------------------------------------------------------

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        patience_count = 0

        torch.save(
            model.state_dict(),
            best_model_path
        )

        print("✓ Best model saved.")

    else:

        patience_count += 1

    # --------------------------------------------------------
    # EARLY STOPPING
    # --------------------------------------------------------

    if patience_count >= patience:

        print("Early stopping.")

        break
# ============================================================
# 13. LOAD BEST MODEL
# ============================================================

model.load_state_dict(
    torch.load(
        best_model_path,
        map_location=device
    )
)

model.eval()

print("Best ConvNeXt-Tiny + MLFF model loaded.")
# ============================================================
# 14. EVALUATION
# ============================================================

actual = []

predicted = []

probabilities = []

with torch.no_grad():

    for images, labels_batch in val_loader:

        images = images.to(device)

        outputs = model(images)

        probs = torch.softmax(
            outputs,
            dim=1
        )

        predictions = torch.argmax(
            outputs,
            dim=1
        )

        actual.extend(
            labels_batch.numpy()
        )

        predicted.extend(
            predictions.cpu().numpy()
        )

        probabilities.extend(
            probs.cpu().numpy()
        )

actual = np.array(actual)

predicted = np.array(predicted)

probabilities = np.array(probabilities)

print("Predictions generated successfully!")
# ============================================================
# 15. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    actual,
    predicted
)

print("Confusion Matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

accuracy = accuracy_score(
    actual,
    predicted
)

sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0

specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

precision = precision_score(
    actual,
    predicted,
    zero_division=0
)

f1 = f1_score(
    actual,
    predicted,
    zero_division=0
)

print("\n==============================")
print("MODEL PERFORMANCE")
print("==============================")

print(
    f"Accuracy    : {accuracy * 100:.2f}%"
)

print(
    f"Sensitivity : {sensitivity * 100:.2f}%"
)

print(
    f"Specificity : {specificity * 100:.2f}%"
)

print(
    f"Precision   : {precision * 100:.2f}%"
)

print(
    f"Recall      : {sensitivity * 100:.2f}%"
)

print(
    f"F1-Score    : {f1 * 100:.2f}%"
)
# ============================================================
# 16. CONFUSION MATRIX PLOT
# ============================================================

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Normal",
        "Glaucoma"
    ],
    yticklabels=[
        "Normal",
        "Glaucoma"
    ]
)

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.title(
    "ConvNeXt-Tiny + MLFF\n"
    "Confusion Matrix"
)

plt.tight_layout()

plt.show()
# ============================================================
# 17. MULTI-LEVEL FEATURE ANALYSIS
# ============================================================

model.eval()

# Get one validation image
image, label = val_dataset[0]

image_input = image.unsqueeze(0).to(device)

with torch.no_grad():

    features = model.backbone(
        image_input
    )

    f1 = model.conv1(
        features[0]
    )

    f2 = model.conv2(
        features[1]
    )

    f3 = model.conv3(
        features[2]
    )

    target_size = f3.shape[2:]

    f1 = F.interpolate(
        f1,
        size=target_size,
        mode="bilinear",
        align_corners=False
    )

    f2 = F.interpolate(
        f2,
        size=target_size,
        mode="bilinear",
        align_corners=False
    )

    fused = torch.cat(
        [f1, f2, f3],
        dim=1
    )

    fused = model.fusion(
        fused
    )


feature_maps = {
    "Level 1": f1,
    "Level 2": f2,
    "Level 3": f3,
    "Fused MLFF": fused
}

results = []

for name, feature in feature_maps.items():

    activation = feature.abs()

    mean_activation = (
        activation.mean().item()
    )

    std_activation = (
        activation.std().item()
    )

    max_activation = (
        activation.max().item()
    )

    threshold = activation.mean()

    active_percentage = (
        (activation > threshold)
        .float()
        .mean()
        .item()
        * 100
    )

    results.append([
        name,
        mean_activation,
        std_activation,
        max_activation,
        active_percentage
    ])


print("\nQuantitative Feature Analysis")
print("-" * 75)

print(
    f"{'Feature':<15}"
    f"{'Mean':<12}"
    f"{'Std':<12}"
    f"{'Max':<12}"
    f"{'Active %':<12}"
)

print("-" * 75)

for row in results:

    print(
        f"{row[0]:<15}"
        f"{row[1]:<12.4f}"
        f"{row[2]:<12.4f}"
        f"{row[3]:<12.4f}"
        f"{row[4]:<12.2f}"
    )
# ============================================================
# 18. FEATURE MAP VISUALIZATION
# ============================================================

fig, axes = plt.subplots(
    1,
    4,
    figsize=(16, 4)
)

for ax, (name, feature) in zip(
    axes,
    feature_maps.items()
):

    activation_map = (
        feature[0]
        .abs()
        .mean(dim=0)
        .cpu()
        .numpy()
    )

    ax.imshow(
        activation_map,
        cmap="viridis"
    )

    ax.set_title(name)

    ax.axis("off")

plt.suptitle(
    "ConvNeXt-Tiny Multi-Level Feature Fusion"
)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# CONVNEXT-TINY + MLFF
# GLAUCOMA CLASSIFICATION
# + GRAD-CAM
# + PRETRAINED VLM (BLIP)
#
# SINGLE CELL - TRAINING + EXPLAINABILITY SETUP
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q transformers accelerate


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader

from torchvision import transforms
from torchvision import models
from torchvision.models import ConvNeXt_Tiny_Weights

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# VLM
from transformers import (
    BlipProcessor,
    BlipForConditionalGeneration
)

from google.colab import drive


# ============================================================
# 3. MOUNT GOOGLE DRIVE
# ============================================================

drive.mount(
    "/content/drive"
)


# ============================================================
# 4. PATHS
# ============================================================

DATA_DIR = (
    "/content/drive/MyDrive/"
    "Database/Database/Images"
)

RESULTS_DIR = (
    "/content/drive/MyDrive/"
    "ConvNeXt_MLFF_Results"
)

CHECKPOINT_DIR = os.path.join(
    RESULTS_DIR,
    "checkpoints"
)

DEMO_DIR = os.path.join(
    RESULTS_DIR,
    "demo_results"
)

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

os.makedirs(
    CHECKPOINT_DIR,
    exist_ok=True
)

os.makedirs(
    DEMO_DIR,
    exist_ok=True
)


print("Dataset path:")
print(DATA_DIR)

print("\nResults path:")
print(RESULTS_DIR)


# ============================================================
# 5. RANDOM SEED
# ============================================================

SEED = 42

random.seed(SEED)

np.random.seed(SEED)

torch.manual_seed(SEED)

torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True

torch.backends.cudnn.benchmark = False


# ============================================================
# 6. DEVICE
# ============================================================

DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print(
    "\nDevice:",
    DEVICE
)


# ============================================================
# 7. CREATE DATAFRAME
# ============================================================

VALID_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
)

image_paths = []

for filename in os.listdir(DATA_DIR):

    if filename.lower().endswith(
        VALID_EXTENSIONS
    ):

        image_paths.append(
            os.path.join(
                DATA_DIR,
                filename
            )
        )


print(
    "\nTotal images found:",
    len(image_paths)
)


data = []

for path in image_paths:

    filename = os.path.basename(
        path
    ).lower()


    # ACRIMA labeling rule
    if "_g_" in filename:

        label = 1

        class_name = "Glaucoma"

    else:

        label = 0

        class_name = "Normal"


    data.append({

        "image_path": path,

        "label": label,

        "class_name": class_name

    })


df = pd.DataFrame(
    data
)


print(
    "\nDataset distribution:"
)

print(
    df["class_name"].value_counts()
)


print(
    "\nTotal images:",
    len(df)
)


# ============================================================
# 8. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

train_df, temp_df = train_test_split(

    df,

    test_size=0.20,

    stratify=df["label"],

    random_state=SEED
)


val_df, test_df = train_test_split(

    temp_df,

    test_size=0.50,

    stratify=temp_df["label"],

    random_state=SEED
)


train_df = train_df.reset_index(
    drop=True
)

val_df = val_df.reset_index(
    drop=True
)

test_df = test_df.reset_index(
    drop=True
)


print(
    "\n=============================="
)

print(
    "DATASET SPLIT"
)

print(
    "=============================="
)

print(
    "Train:",
    len(train_df)
)

print(
    "Validation:",
    len(val_df)
)

print(
    "Test:",
    len(test_df)
)


# ============================================================
# 9. IMAGE TRANSFORMS
# ============================================================

train_transform = transforms.Compose([

    transforms.Resize(
        (224, 224)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15
    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[
            0.485,
            0.456,
            0.406
        ],

        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])


val_transform = transforms.Compose([

    transforms.Resize(
        (224, 224)
    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[
            0.485,
            0.456,
            0.406
        ],

        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])


# ============================================================
# 10. DATASET CLASS
# ============================================================

class ACRIMADataset(
    Dataset
):

    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.dataframe = (
            dataframe.reset_index(
                drop=True
            )
        )

        self.transform = transform


    def __len__(
        self
    ):

        return len(
            self.dataframe
        )


    def __getitem__(
        self,
        index
    ):

        row = (
            self.dataframe.iloc[
                index
            ]
        )


        image_path = (
            row["image_path"]
        )

        label = int(
            row["label"]
        )


        image = Image.open(
            image_path
        ).convert(
            "RGB"
        )


        if self.transform:

            image = self.transform(
                image
            )


        return (

            image,

            torch.tensor(
                label,
                dtype=torch.long
            )
        )


# ============================================================
# 11. CREATE DATASETS
# ============================================================

train_dataset = ACRIMADataset(

    train_df,

    train_transform
)


val_dataset = ACRIMADataset(

    val_df,

    val_transform
)


test_dataset = ACRIMADataset(

    test_df,

    val_transform
)


# ============================================================
# 12. CREATE DATALOADERS
# ============================================================

BATCH_SIZE = 16


train_loader = DataLoader(

    train_dataset,

    batch_size=BATCH_SIZE,

    shuffle=True,

    num_workers=2,

    pin_memory=True
)


val_loader = DataLoader(

    val_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=2,

    pin_memory=True
)


test_loader = DataLoader(

    test_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=2,

    pin_memory=True
)


print(
    "\nDataLoaders created."
)


# ============================================================
# 13. MULTI-LEVEL FEATURE FUSION
# ============================================================

class MLFF(
    nn.Module
):

    def __init__(
        self,
        channels,
        fusion_channels=128
    ):

        super().__init__()


        self.projections = (
            nn.ModuleList([

                nn.Sequential(

                    nn.Conv2d(

                        c,

                        fusion_channels,

                        kernel_size=1,

                        bias=False
                    ),

                    nn.BatchNorm2d(
                        fusion_channels
                    ),

                    nn.ReLU(
                        inplace=True
                    )
                )

                for c in channels
            ])
        )


    def forward(
        self,
        features
    ):

        projected = []


        # Spatial size of deepest feature
        target_size = (
            features[-1].shape[-2:]
        )


        for feature, projection in zip(

            features,

            self.projections

        ):

            x = projection(
                feature
            )


            if x.shape[-2:] != target_size:

                x = F.adaptive_avg_pool2d(

                    x,

                    target_size
                )


            projected.append(
                x
            )


        # Concatenate all feature levels
        fused = torch.cat(

            projected,

            dim=1
        )


        return fused


# ============================================================
# 14. CONVNEXT-TINY + MLFF MODEL
# ============================================================

class ConvNeXtTinyMLFF(
    nn.Module
):

    def __init__(
        self,
        num_classes=2,
        pretrained=True
    ):

        super().__init__()


        # ----------------------------------------------------
        # PRETRAINED CONVNEXT-TINY
        # ----------------------------------------------------

        if pretrained:

            weights = (
                ConvNeXt_Tiny_Weights.DEFAULT
            )

        else:

            weights = None


        backbone = (
            models.convnext_tiny(
                weights=weights
            )
        )


        self.backbone = backbone


        # ----------------------------------------------------
        # CONVNEXT-TINY FEATURE CHANNELS
        # ----------------------------------------------------

        self.feature_channels = [

            96,

            192,

            384,

            768
        ]


        # ----------------------------------------------------
        # MLFF
        # ----------------------------------------------------

        self.mlff = MLFF(

            channels=self.feature_channels,

            fusion_channels=128
        )


        # 4 feature levels × 128 channels
        fused_channels = 512


        # ----------------------------------------------------
        # CLASSIFIER
        # ----------------------------------------------------

        self.classifier = nn.Sequential(

            nn.AdaptiveAvgPool2d(
                (1, 1)
            ),

            nn.Flatten(),

            nn.Dropout(
                p=0.40
            ),

            nn.Linear(

                fused_channels,

                128
            ),

            nn.ReLU(
                inplace=True
            ),

            nn.Dropout(
                p=0.30
            ),

            nn.Linear(

                128,

                num_classes
            )
        )


    def forward(

        self,

        x,

        return_features=False

    ):

        features = []


        # ----------------------------------------------------
        # EXTRACT MULTI-LEVEL CONVNEXT FEATURES
        # ----------------------------------------------------

        for i, layer in enumerate(

            self.backbone.features

        ):

            x = layer(x)


            # ConvNeXt-Tiny stages
            if i in [
                1,
                3,
                5,
                7
            ]:

                features.append(
                    x
                )


        # ----------------------------------------------------
        # MLFF
        # ----------------------------------------------------

        fused = self.mlff(
            features
        )


        # ----------------------------------------------------
        # CLASSIFICATION
        # ----------------------------------------------------

        logits = self.classifier(
            fused
        )


        if return_features:

            return (

                logits,

                features,

                fused
            )


        return logits


# ============================================================
# 15. CREATE MODEL
# ============================================================

model = ConvNeXtTinyMLFF(

    num_classes=2,

    pretrained=True
)


model = model.to(
    DEVICE
)


print(
    "\nModel created successfully."
)


print(
    model
)


# ============================================================
# 16. FREEZE EARLY CONVNEXT LAYERS
# ============================================================

for i, layer in enumerate(

    model.backbone.features

):

    if i <= 3:

        for param in (
            layer.parameters()
        ):

            param.requires_grad = False


print(
    "\nEarly ConvNeXt layers frozen."
)


# ============================================================
# 17. CLASS WEIGHTS
# ============================================================

class_counts = np.bincount(

    train_df["label"].values
)


total_samples = len(
    train_df
)


class_weights = (

    total_samples /

    (
        2 *
        class_counts
    )
)


class_weights = torch.tensor(

    class_weights,

    dtype=torch.float32

).to(
    DEVICE
)


print(
    "\nClass weights:"
)

print(
    class_weights
)


# ============================================================
# 18. LOSS
# ============================================================

criterion = nn.CrossEntropyLoss(

    weight=class_weights,

    label_smoothing=0.05
)


# ============================================================
# 19. OPTIMIZER
# ============================================================

optimizer = optim.AdamW(

    filter(

        lambda p:
        p.requires_grad,

        model.parameters()
    ),

    lr=1e-4,

    weight_decay=1e-4
)


# ============================================================
# 20. LR SCHEDULER
# ============================================================

scheduler = (
    optim.lr_scheduler.ReduceLROnPlateau(

        optimizer,

        mode="max",

        factor=0.5,

        patience=3
    )
)


# ============================================================
# 21. TRAINING FUNCTION
# ============================================================

def train_one_epoch(

    model,

    loader,

    criterion,

    optimizer

):

    model.train()


    running_loss = 0.0

    correct = 0

    total = 0


    for images, labels in loader:


        images = images.to(

            DEVICE,

            non_blocking=True
        )


        labels = labels.to(

            DEVICE,

            non_blocking=True
        )


        optimizer.zero_grad()


        outputs = model(
            images
        )


        loss = criterion(

            outputs,

            labels
        )


        loss.backward()


        torch.nn.utils.clip_grad_norm_(

            model.parameters(),

            max_norm=5.0
        )


        optimizer.step()


        running_loss += (

            loss.item() *

            images.size(0)
        )


        predictions = (

            outputs.argmax(
                dim=1
            )
        )


        correct += (

            predictions == labels

        ).sum().item()


        total += (
            labels.size(0)
        )


    epoch_loss = (

        running_loss /

        total
    )


    epoch_accuracy = (

        correct /

        total
    )


    return (

        epoch_loss,

        epoch_accuracy
    )


# ============================================================
# 22. VALIDATION FUNCTION
# ============================================================

@torch.no_grad()
def evaluate(

    model,

    loader,

    criterion

):

    model.eval()


    running_loss = 0.0

    total = 0


    all_labels = []

    all_predictions = []

    all_probabilities = []


    for images, labels in loader:


        images = images.to(

            DEVICE,

            non_blocking=True
        )


        labels = labels.to(

            DEVICE,

            non_blocking=True
        )


        outputs = model(
            images
        )


        loss = criterion(

            outputs,

            labels
        )


        running_loss += (

            loss.item() *

            images.size(0)
        )


        probabilities = (

            torch.softmax(

                outputs,

                dim=1
            )
        )


        predictions = (

            outputs.argmax(

                dim=1
            )
        )


        all_labels.extend(

            labels.cpu().numpy()
        )


        all_predictions.extend(

            predictions.cpu().numpy()
        )


        all_probabilities.extend(

            probabilities[:, 1]

            .cpu()

            .numpy()
        )


        total += (

            labels.size(0)
        )


    loss = (

        running_loss /

        total
    )


    accuracy = accuracy_score(

        all_labels,

        all_predictions
    )


    precision = precision_score(

        all_labels,

        all_predictions,

        zero_division=0
    )


    recall = recall_score(

        all_labels,

        all_predictions,

        zero_division=0
    )


    f1 = f1_score(

        all_labels,

        all_predictions,

        zero_division=0
    )


    auc = roc_auc_score(

        all_labels,

        all_probabilities
    )


    return {

        "loss": loss,

        "accuracy": accuracy,

        "precision": precision,

        "recall": recall,

        "f1": f1,

        "auc": auc
    }


# ============================================================
# 23. TRAIN MODEL
# ============================================================

NUM_EPOCHS = 30

best_val_auc = 0.0

patience = 7

patience_counter = 0

history = []


print(
    "\n=============================="
)

print(
    "STARTING TRAINING"
)

print(
    "=============================="
)


for epoch in range(
    NUM_EPOCHS
):


    train_loss, train_acc = (
        train_one_epoch(

            model,

            train_loader,

            criterion,

            optimizer
        )
    )


    val_metrics = evaluate(

        model,

        val_loader,

        criterion
    )


    scheduler.step(

        val_metrics["auc"]
    )


    print(
        f"\nEpoch "
        f"[{epoch+1}/{NUM_EPOCHS}]"
    )


    print(

        f"Train Loss: "
        f"{train_loss:.4f} | "

        f"Train Acc: "
        f"{train_acc:.4f}"
    )


    print(

        f"Val Loss: "
        f"{val_metrics['loss']:.4f} | "

        f"Val Acc: "
        f"{val_metrics['accuracy']:.4f}"
    )


    print(

        f"Val Precision: "
        f"{val_metrics['precision']:.4f} | "

        f"Val Recall: "
        f"{val_metrics['recall']:.4f} | "

        f"Val F1: "
        f"{val_metrics['f1']:.4f} | "

        f"Val AUC: "
        f"{val_metrics['auc']:.4f}"
    )


    history.append({

        "epoch":
            epoch + 1,

        "train_loss":
            train_loss,

        "train_accuracy":
            train_acc,

        "val_loss":
            val_metrics["loss"],

        "val_accuracy":
            val_metrics["accuracy"],

        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_auc":
            val_metrics["auc"]
    })


    # --------------------------------------------------------
    # SAVE BEST MODEL
    # --------------------------------------------------------

    if (

        val_metrics["auc"]

        >

        best_val_auc

    ):


        best_val_auc = (
            val_metrics["auc"]
        )


        patience_counter = 0


        checkpoint_path = os.path.join(

            CHECKPOINT_DIR,

            "best_convnext_tiny_mlff.pth"
        )


        torch.save({

            "model_state_dict":
                model.state_dict(),

            "best_val_auc":
                best_val_auc,

            "epoch":
                epoch + 1

        }, checkpoint_path)


        print(
            "✓ Best model saved."
        )


    else:

        patience_counter += 1


    # --------------------------------------------------------
    # EARLY STOPPING
    # --------------------------------------------------------

    if (

        patience_counter >=
        patience

    ):

        print(
            "\nEarly stopping."
        )

        break


print(
    "\nTraining completed."
)


print(
    "Best validation AUC:",
    best_val_auc
)


# ============================================================
# 24. LOAD BEST MODEL
# ============================================================

checkpoint_path = os.path.join(

    CHECKPOINT_DIR,

    "best_convnext_tiny_mlff.pth"
)


checkpoint = torch.load(

    checkpoint_path,

    map_location=DEVICE
)


model.load_state_dict(

    checkpoint[
        "model_state_dict"
    ]
)


model = model.to(
    DEVICE
)


model.eval()


print(
    "\nBest model loaded successfully."
)


print(
    "Best validation AUC:",
    checkpoint[
        "best_val_auc"
    ]
)


# ============================================================
# 25. FINAL TEST EVALUATION
# ============================================================

test_metrics = evaluate(

    model,

    test_loader,

    criterion
)


print(
    "\n=============================="
)

print(
    "FINAL TEST RESULTS"
)

print(
    "=============================="
)


for key, value in (
    test_metrics.items()
):

    print(

        f"{key.upper():12s}: "
        f"{value:.4f}"
    )


# ============================================================
# 26. CONFUSION MATRIX
# ============================================================

@torch.no_grad()
def get_predictions(

    model,

    loader

):

    model.eval()


    labels_all = []

    preds_all = []


    for images, labels in loader:


        images = images.to(
            DEVICE
        )


        outputs = model(
            images
        )


        preds = outputs.argmax(

            dim=1
        )


        labels_all.extend(

            labels.numpy()
        )


        preds_all.extend(

            preds.cpu().numpy()
        )


    return (

        labels_all,

        preds_all
    )


y_true, y_pred = get_predictions(

    model,

    test_loader
)


cm = confusion_matrix(

    y_true,

    y_pred
)


print(
    "\nConfusion Matrix:"
)

print(cm)


print(
    "\nClassification Report:"
)

print(

    classification_report(

        y_true,

        y_pred,

        target_names=[

            "Normal",

            "Glaucoma"
        ]
    )
)


# ============================================================
# 27. TRAINING CURVES
# ============================================================

history_df = pd.DataFrame(
    history
)


plt.figure(
    figsize=(8, 5)
)


plt.plot(

    history_df[
        "train_accuracy"
    ],

    label="Train Accuracy"
)


plt.plot(

    history_df[
        "val_accuracy"
    ],

    label="Validation Accuracy"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "ConvNeXt-Tiny + MLFF Accuracy"
)

plt.legend()

plt.grid()

plt.show()


plt.figure(
    figsize=(8, 5)
)


plt.plot(

    history_df[
        "train_loss"
    ],

    label="Train Loss"
)


plt.plot(

    history_df[
        "val_loss"
    ],

    label="Validation Loss"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.title(
    "ConvNeXt-Tiny + MLFF Loss"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 28. GRAD-CAM CLASS
# ============================================================

class GradCAM:

    def __init__(

        self,

        model,

        target_layer

    ):

        self.model = model

        self.target_layer = (
            target_layer
        )

        self.activations = None

        self.gradients = None


        self.forward_handle = (

            target_layer.register_forward_hook(

                self.save_activation
            )
        )


        self.backward_handle = (

            target_layer.register_full_backward_hook(

                self.save_gradient
            )
        )


    def save_activation(

        self,

        module,

        input,

        output

    ):

        self.activations = output


    def save_gradient(

        self,

        module,

        grad_input,

        grad_output

    ):

        self.gradients = (
            grad_output[0]
        )


    def generate(

        self,

        image_tensor,

        target_class

    ):

        self.model.zero_grad()


        logits = self.model(

            image_tensor
        )


        score = logits[

            0,

            target_class

        ]


        score.backward()


        activations = (
            self.activations
        )


        gradients = (
            self.gradients
        )


        # Global average pooling
        weights = gradients.mean(

            dim=(2, 3),

            keepdim=True
        )


        # Weighted activation maps
        cam = (

            weights *

            activations

        ).sum(

            dim=1,

            keepdim=True
        )


        cam = F.relu(
            cam
        )


        # Resize to input resolution
        cam = F.interpolate(

            cam,

            size=(

                224,

                224
            ),

            mode="bilinear",

            align_corners=False
        )


        cam = (

            cam

            .squeeze()

            .detach()

            .cpu()

            .numpy()
        )


        # Normalize
        cam -= cam.min()


        if cam.max() > 0:

            cam /= cam.max()


        return cam


    def close(
        self
    ):

        self.forward_handle.remove()

        self.backward_handle.remove()


# ============================================================
# 29. GRAD-CAM HELPER FUNCTIONS
# ============================================================

def denormalize_image(

    tensor

):

    image = tensor.clone()


    mean = torch.tensor([

        0.485,

        0.456,

        0.406

    ]).view(

        3,

        1,

        1
    )


    std = torch.tensor([

        0.229,

        0.224,

        0.225

    ]).view(

        3,

        1,

        1
    )


    image = (

        image.cpu() *

        std

    ) + mean


    image = torch.clamp(

        image,

        0,

        1
    )


    image = image.permute(

        1,

        2,

        0

    ).numpy()


    return image


def create_gradcam_overlay(

    original_image,

    cam

):

    original = np.array(

        original_image.resize(

            (224, 224)
        )
    )


    heatmap = np.uint8(

        255 * cam
    )


    heatmap = cv2.applyColorMap(

        heatmap,

        cv2.COLORMAP_JET
    )


    heatmap = cv2.cvtColor(

        heatmap,

        cv2.COLOR_BGR2RGB
    )


    overlay = cv2.addWeighted(

        original,

        0.55,

        heatmap,

        0.45,

        0
    )


    return overlay


# ============================================================
# 30. PREPARE GRAD-CAM TARGET LAYER
# ============================================================

# Final ConvNeXt feature stage
GRADCAM_TARGET_LAYER = (

    model.backbone.features[7]
)


print(
    "\nGrad-CAM target layer:"
)

print(
    GRADCAM_TARGET_LAYER
)


# ============================================================
# 31. LOAD PRETRAINED VLM - BLIP
# ============================================================

print(
    "\n=============================="
)

print(
    "LOADING PRETRAINED VLM"
)

print(
    "=============================="
)


VLM_DEVICE = (

    "cuda"

    if torch.cuda.is_available()

    else "cpu"
)


print(
    "VLM Device:",
    VLM_DEVICE
)


# ------------------------------------------------------------
# BLIP PROCESSOR
# ------------------------------------------------------------

vlm_processor = (

    BlipProcessor.from_pretrained(

        "Salesforce/blip-image-captioning-base"
    )
)


# ------------------------------------------------------------
# BLIP MODEL
# ------------------------------------------------------------

vlm_model = (

    BlipForConditionalGeneration.from_pretrained(

        "Salesforce/blip-image-captioning-base"
    )
)


vlm_model = vlm_model.to(

    VLM_DEVICE
)


vlm_model.eval()


print(
    "\n✓ Pretrained BLIP VLM loaded."
)


# ============================================================
# 32. VLM DESCRIPTION FUNCTION
# ============================================================

def generate_vlm_description(

    image

):

    prompt = (

        "A medical retinal fundus "
        "photograph showing"
    )


    inputs = vlm_processor(

        images=image,

        text=prompt,

        return_tensors="pt"
    )


    inputs = {

        key: value.to(
            VLM_DEVICE
        )

        for key, value
        in inputs.items()
    }


    with torch.no_grad():

        generated_ids = (

            vlm_model.generate(

                **inputs,

                max_new_tokens=50,

                num_beams=5
            )
        )


    generated_text = (

        vlm_processor.decode(

            generated_ids[0],

            skip_special_tokens=True
        )
    )


    return generated_text


# ============================================================
# 33. FINAL STATUS
# ============================================================

print(
    "\n"
    "===================================================="
)

print(
    "COMPLETE PIPELINE READY"
)

print(
    "===================================================="
)

print(
    "✓ ACRIMA dataset loaded"
)

print(
    "✓ Train / validation / test split created"
)

print(
    "✓ ConvNeXt-Tiny backbone created"
)

print(
    "✓ Multi-Level Feature Fusion (MLFF) added"
)

print(
    "✓ Glaucoma / Normal classifier trained"
)

print(
    "✓ Best checkpoint loaded"
)

print(
    "✓ Test evaluation completed"
)

print(
    "✓ Grad-CAM initialized"
)

print(
    "✓ Pretrained BLIP VLM loaded"
)

print(
    "✓ No VLM fine-tuning performed"
)

print(
    "\nNext step:"
)

print(
    "Use the separate demo cell to upload "
    "a new fundus image and obtain:"
)

print(
    "Prediction + Probability + Grad-CAM + VLM description"
)

print(
    "===================================================="
)
```

```python
# ============================================================
# ACRIMA + RIM-ONE DL
# ConvNeXt-Tiny + Multi-Level Feature Fusion (MLFF)
# Data Augmentation + Grad-CAM + Pretrained BLIP VLM
#
# NO VLM FINE-TUNING
# NO DEMO / SINGLE IMAGE PREDICTION IN THIS CELL
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

!pip install -q transformers accelerate


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import random
import zipfile
import shutil
import copy
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from torchvision.models import ConvNeXt_Tiny_Weights

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import matplotlib.pyplot as plt

from transformers import (
    BlipProcessor,
    BlipForConditionalGeneration
)

print("PyTorch version:", torch.__version__)


# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False


# ============================================================
# 4. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount('/content/drive')


# ============================================================
# 5. PATHS
# ============================================================

# ------------------------------------------------------------
# ACRIMA
# ------------------------------------------------------------

ACRIMA_DIR = "/content/drive/MyDrive/Database/Database/Images"


# ------------------------------------------------------------
# RIM-ONE ZIP
#
# Upload your RIM-ONE_DL_images.zip to Google Drive.
# Change this path only if you put the ZIP somewhere else.
# ------------------------------------------------------------



RIM_EXTRACT_DIR = "/content/drive/MyDrive/RIM-ONE_DL_images"


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

RESULTS_DIR = "/content/drive/MyDrive/ConvNeXt_MLFF_Combined_Results"

CHECKPOINT_DIR = os.path.join(
    RESULTS_DIR,
    "checkpoints"
)

GRADCAM_DIR = os.path.join(
    RESULTS_DIR,
    "gradcam"
)

os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(GRADCAM_DIR, exist_ok=True)


CHECKPOINT_PATH = os.path.join(
    CHECKPOINT_DIR,
    "best_convnext_tiny_mlff_combined.pth"
)


print("\nResults directory:")
print(RESULTS_DIR)



# ============================================================
# 7. FIND ACTUAL RIM-ONE ROOT
# ============================================================

# Your ZIP contains:
#
# RIM-ONE_DL_images/
#     partitioned_by_hospital/
#
# Therefore this should exist:

RIM_ROOT = os.path.join(
    RIM_EXTRACT_DIR,
    "partitioned_by_hospital"
)

if not os.path.exists(RIM_ROOT):

    raise FileNotFoundError(
        f"RIM-ONE root not found:\n{RIM_ROOT}"
    )

print("\nRIM-ONE root:")
print(RIM_ROOT)


# ============================================================
# 8. CREATE ACRIMA DATAFRAME
# ============================================================

def collect_acrima_images(root_dir):

    records = []

    if not os.path.exists(root_dir):
        raise FileNotFoundError(
            f"ACRIMA directory not found:\n{root_dir}"
        )

    for filename in os.listdir(root_dir):

        if not filename.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")
        ):
            continue

        image_path = os.path.join(
            root_dir,
            filename
        )

        # ACRIMA labeling rule:
        # "_g_" -> glaucoma
        # otherwise -> normal

        if "_g_" in filename.lower():
            label = 1
            class_name = "glaucoma"
        else:
            label = 0
            class_name = "normal"

        records.append({
            "image_path": image_path,
            "label": label,
            "class_name": class_name,
            "dataset": "ACRIMA"
        })

    return pd.DataFrame(records)


acrima_df = collect_acrima_images(
    ACRIMA_DIR
)


print("\nACRIMA:")
print(acrima_df["class_name"].value_counts())
print("Total:", len(acrima_df))


# ============================================================
# 9. CREATE RIM-ONE DATAFRAME
# ============================================================

def collect_rimone_images(root_dir):

    records = []

    for split_name in [
        "training_set",
        "test_set"
    ]:

        split_dir = os.path.join(
            root_dir,
            split_name
        )

        if not os.path.exists(split_dir):
            continue

        for class_name, label in [
            ("normal", 0),
            ("glaucoma", 1)
        ]:

            class_dir = os.path.join(
                split_dir,
                class_name
            )

            if not os.path.exists(class_dir):
                continue

            for filename in os.listdir(class_dir):

                if not filename.lower().endswith(
                    (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")
                ):
                    continue

                image_path = os.path.join(
                    class_dir,
                    filename
                )

                records.append({
                    "image_path": image_path,
                    "label": label,
                    "class_name": class_name,
                    "dataset": "RIM-ONE"
                })

    return pd.DataFrame(records)


rim_df = collect_rimone_images(
    RIM_ROOT
)


print("\nRIM-ONE:")
print(rim_df["class_name"].value_counts())
print("Total:", len(rim_df))


# ============================================================
# 10. COMBINE ACRIMA + RIM-ONE
# ============================================================

combined_df = pd.concat(
    [
        acrima_df,
        rim_df
    ],
    ignore_index=True
)


# Shuffle once
combined_df = combined_df.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)


print("\n======================================")
print("COMBINED DATASET")
print("======================================")

print("\nTotal images:", len(combined_df))

print("\nClass distribution:")
print(
    combined_df["class_name"].value_counts()
)

print("\nDataset distribution:")
print(
    combined_df["dataset"].value_counts()
)

print("\nDataset + class:")
print(
    pd.crosstab(
        combined_df["dataset"],
        combined_df["class_name"]
    )
)


# ============================================================
# 11. REMOVE EXACT DUPLICATE FILE PATHS
# ============================================================

before = len(combined_df)

combined_df = combined_df.drop_duplicates(
    subset=["image_path"]
).reset_index(drop=True)

after = len(combined_df)

print(
    f"\nRemoved duplicate paths: {before - after}"
)


# ============================================================
# 12. CHECK THAT IMAGES CAN BE OPENED
# ============================================================

print("\nChecking image files...")

bad_images = []

for path in combined_df["image_path"]:

    try:

        with Image.open(path) as img:
            img.verify()

    except Exception:

        bad_images.append(path)


if len(bad_images) > 0:

    print(
        f"Found {len(bad_images)} invalid images."
    )

    combined_df = combined_df[
        ~combined_df["image_path"].isin(bad_images)
    ].reset_index(drop=True)

else:

    print("All images are readable.")


# ============================================================
# 13. STRATIFIED TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# 70% TRAIN
# 15% VALIDATION
# 15% TEST

train_df, temp_df = train_test_split(
    combined_df,
    test_size=0.30,
    stratify=combined_df["label"],
    random_state=SEED
)


val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=SEED
)


train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)


print("\n======================================")
print("DATASET SPLIT")
print("======================================")

print(
    f"Train      : {len(train_df)}"
)

print(
    f"Validation : {len(val_df)}"
)

print(
    f"Test       : {len(test_df)}"
)


print("\nTRAIN distribution:")
print(
    train_df["class_name"].value_counts()
)


print("\nVALIDATION distribution:")
print(
    val_df["class_name"].value_counts()
)


print("\nTEST distribution:")
print(
    test_df["class_name"].value_counts()
)


print("\nTrain dataset source:")
print(
    train_df["dataset"].value_counts()
)

print("\nValidation dataset source:")
print(
    val_df["dataset"].value_counts()
)

print("\nTest dataset source:")
print(
    test_df["dataset"].value_counts()
)


# ============================================================
# 14. SAVE SPLIT INFORMATION
# ============================================================

train_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "train_split.csv"
    ),
    index=False
)

val_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "validation_split.csv"
    ),
    index=False
)

test_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "test_split.csv"
    ),
    index=False
)


# ============================================================
# 15. DEVICE
# ============================================================

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nDevice:", DEVICE)


# ============================================================
# 16. DATA AUGMENTATION
# ============================================================

# IMPORTANT:
#
# Augmentation is used ONLY for TRAINING.
#
# Validation and TEST do NOT use augmentation.

train_transform = transforms.Compose([

    transforms.Resize((256, 256)),

    transforms.RandomResizedCrop(
        224,
        scale=(0.80, 1.0),
        ratio=(0.95, 1.05)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=15
    ),

    transforms.ColorJitter(
        brightness=0.20,
        contrast=0.20,
        saturation=0.10
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# Validation/Test = NO augmentation

eval_transform = transforms.Compose([

    transforms.Resize((224, 224)),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ============================================================
# 17. DATASET CLASS
# ============================================================

class FundusDataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.dataframe = dataframe.reset_index(
            drop=True
        )

        self.transform = transform


    def __len__(self):

        return len(self.dataframe)


    def __getitem__(self, idx):

        row = self.dataframe.iloc[idx]

        image_path = row["image_path"]

        label = int(row["label"])

        image = Image.open(
            image_path
        ).convert("RGB")


        if self.transform is not None:

            image = self.transform(image)


        return (
            image,
            torch.tensor(
                label,
                dtype=torch.long
            )
        )


# ============================================================
# 18. CREATE DATASETS
# ============================================================

train_dataset = FundusDataset(
    train_df,
    train_transform
)

val_dataset = FundusDataset(
    val_df,
    eval_transform
)

test_dataset = FundusDataset(
    test_df,
    eval_transform
)


# ============================================================
# 19. DATALOADERS
# ============================================================

BATCH_SIZE = 16

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


print("\nDataLoaders created.")


# ============================================================
# 20. CONVNEXT-TINY + MLFF
# ============================================================

class ConvNeXtTinyMLFF(nn.Module):

    def __init__(self):

        super().__init__()


        # ----------------------------------------------------
        # PRETRAINED CONVNEXT-TINY
        # ----------------------------------------------------

        weights = ConvNeXt_Tiny_Weights.DEFAULT

        backbone = models.convnext_tiny(
            weights=weights
        )

        self.backbone = backbone


        # ----------------------------------------------------
        # MULTI-LEVEL FEATURES
        #
        # ConvNeXt Tiny:
        #
        # Level 1 -> 96
        # Level 2 -> 192
        # Level 3 -> 384
        # Level 4 -> 768
        # ----------------------------------------------------

        self.proj1 = nn.Sequential(
            nn.Conv2d(
                96,
                128,
                kernel_size=1
            ),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )

        self.proj2 = nn.Sequential(
            nn.Conv2d(
                192,
                128,
                kernel_size=1
            ),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )

        self.proj3 = nn.Sequential(
            nn.Conv2d(
                384,
                128,
                kernel_size=1
            ),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )

        self.proj4 = nn.Sequential(
            nn.Conv2d(
                768,
                128,
                kernel_size=1
            ),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )


        # ----------------------------------------------------
        # MLFF
        # ----------------------------------------------------

        self.fusion = nn.Sequential(

            nn.Conv2d(
                512,
                256,
                kernel_size=1
            ),

            nn.BatchNorm2d(256),

            nn.ReLU(inplace=True),

            nn.Conv2d(
                256,
                128,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(128),

            nn.ReLU(inplace=True)
        )


        # ----------------------------------------------------
        # CLASSIFIER
        # ----------------------------------------------------

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Dropout(0.50),

            nn.Linear(
                128,
                128
            ),

            nn.ReLU(inplace=True),

            nn.Dropout(0.40),

            nn.Linear(
                128,
                2
            )
        )


    def forward(
        self,
        x,
        return_features=False
    ):

        # ----------------------------------------------------
        # EXTRACT CONVNEXT MULTI-LEVEL FEATURES
        # ----------------------------------------------------

        features = []

        for i, layer in enumerate(
            self.backbone.features
        ):

            x = layer(x)

            if i in [1, 3, 5, 7]:

                features.append(x)


        f1 = self.proj1(features[0])
        f2 = self.proj2(features[1])
        f3 = self.proj3(features[2])
        f4 = self.proj4(features[3])


        # ----------------------------------------------------
        # UPSAMPLE ALL FEATURES TO f1 SIZE
        # ----------------------------------------------------

        target_size = f1.shape[-2:]


        f2 = nn.functional.interpolate(
            f2,
            size=target_size,
            mode="bilinear",
            align_corners=False
        )

        f3 = nn.functional.interpolate(
            f3,
            size=target_size,
            mode="bilinear",
            align_corners=False
        )

        f4 = nn.functional.interpolate(
            f4,
            size=target_size,
            mode="bilinear",
            align_corners=False
        )


        # ----------------------------------------------------
        # MULTI-LEVEL FEATURE FUSION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                f1,
                f2,
                f3,
                f4
            ],
            dim=1
        )


        fused = self.fusion(
            fused
        )


        # ----------------------------------------------------
        # CLASSIFICATION
        # ----------------------------------------------------

        pooled = self.pool(
            fused
        )

        logits = self.classifier(
            pooled
        )


        if return_features:

            return logits, fused

        return logits


# ============================================================
# 21. CREATE MODEL
# ============================================================

model = ConvNeXtTinyMLFF()

model = model.to(DEVICE)


print("\nModel created:")
print(model)


# ============================================================
# 22. PARTIAL FREEZING
# ============================================================

# Freeze only the earliest ConvNeXt stage.
#
# More layers remain trainable so the model can
# adapt to ACRIMA + RIM-ONE.

for i, layer in enumerate(
    model.backbone.features
):

    if i <= 1:

        for param in layer.parameters():

            param.requires_grad = False


# Count trainable parameters

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

total_params = sum(
    p.numel()
    for p in model.parameters()
)

print(
    f"\nTrainable parameters: {trainable_params:,}"
)

print(
    f"Total parameters: {total_params:,}"
)


# ============================================================
# 23. CLASS WEIGHTS
# ============================================================

class_counts = (
    train_df["label"]
    .value_counts()
    .sort_index()
)

num_normal = class_counts.get(
    0,
    1
)

num_glaucoma = class_counts.get(
    1,
    1
)

total_train = (
    num_normal +
    num_glaucoma
)

weight_normal = (
    total_train /
    (2 * num_normal)
)

weight_glaucoma = (
    total_train /
    (2 * num_glaucoma)
)

class_weights = torch.tensor(
    [
        weight_normal,
        weight_glaucoma
    ],
    dtype=torch.float32
).to(DEVICE)


print("\nClass weights:")
print(class_weights)


# ============================================================
# 24. LOSS FUNCTION
# ============================================================

criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=0.05
)


# ============================================================
# 25. OPTIMIZER
# ============================================================

optimizer = optim.AdamW(
    filter(
        lambda p: p.requires_grad,
        model.parameters()
    ),
    lr=3e-5,
    weight_decay=1e-3
)


# ============================================================
# 26. LEARNING RATE SCHEDULER
# ============================================================

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2
)


# ============================================================
# 27. TRAINING FUNCTION
# ============================================================

def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device
):

    model.train()

    running_loss = 0.0

    all_preds = []
    all_labels = []


    for images, labels in loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        optimizer.zero_grad()


        logits = model(
            images
        )


        loss = criterion(
            logits,
            labels
        )


        loss.backward()


        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )


        optimizer.step()


        running_loss += (
            loss.item()
            * images.size(0)
        )


        preds = torch.argmax(
            logits,
            dim=1
        )


        all_preds.extend(
            preds.detach()
            .cpu()
            .numpy()
        )

        all_labels.extend(
            labels.detach()
            .cpu()
            .numpy()
        )


    epoch_loss = (
        running_loss /
        len(loader.dataset)
    )


    epoch_acc = accuracy_score(
        all_labels,
        all_preds
    )


    return epoch_loss, epoch_acc


# ============================================================
# 28. VALIDATION FUNCTION
# ============================================================

@torch.no_grad()
def evaluate(
    model,
    loader,
    criterion,
    device
):

    model.eval()

    running_loss = 0.0

    all_labels = []
    all_preds = []
    all_probs = []


    for images, labels in loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        logits = model(
            images
        )


        loss = criterion(
            logits,
            labels
        )


        running_loss += (
            loss.item()
            * images.size(0)
        )


        probs = torch.softmax(
            logits,
            dim=1
        )[:, 1]


        preds = torch.argmax(
            logits,
            dim=1
        )


        all_labels.extend(
            labels.cpu().numpy()
        )

        all_preds.extend(
            preds.cpu().numpy()
        )

        all_probs.extend(
            probs.cpu().numpy()
        )


    loss_value = (
        running_loss /
        len(loader.dataset)
    )


    accuracy = accuracy_score(
        all_labels,
        all_preds
    )


    precision = precision_score(
        all_labels,
        all_preds,
        zero_division=0
    )


    recall = recall_score(
        all_labels,
        all_preds,
        zero_division=0
    )


    f1 = f1_score(
        all_labels,
        all_preds,
        zero_division=0
    )


    try:

        auc = roc_auc_score(
            all_labels,
            all_probs
        )

    except:

        auc = 0.0


    return {
        "loss": loss_value,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "auc": auc,
        "labels": np.array(all_labels),
        "preds": np.array(all_preds),
        "probs": np.array(all_probs)
    }


# ============================================================
# 29. TRAINING LOOP
# ============================================================

EPOCHS = 30

EARLY_STOPPING_PATIENCE = 7

best_val_loss = float("inf")

best_state = None

patience_counter = 0


history = {
    "train_loss": [],
    "train_acc": [],
    "val_loss": [],
    "val_acc": [],
    "val_auc": []
}


print("\n======================================")
print("STARTING TRAINING")
print("======================================")


for epoch in range(
    1,
    EPOCHS + 1
):


    train_loss, train_acc = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        DEVICE
    )


    val_metrics = evaluate(
        model,
        val_loader,
        criterion,
        DEVICE
    )


    val_loss = val_metrics["loss"]

    val_acc = val_metrics["accuracy"]

    val_auc = val_metrics["auc"]


    scheduler.step(
        val_loss
    )


    history["train_loss"].append(
        train_loss
    )

    history["train_acc"].append(
        train_acc
    )

    history["val_loss"].append(
        val_loss
    )

    history["val_acc"].append(
        val_acc
    )

    history["val_auc"].append(
        val_auc
    )


    current_lr = optimizer.param_groups[0]["lr"]


    print(
        f"\nEpoch [{epoch:02d}/{EPOCHS}]"
    )

    print(
        f"Train Loss : {train_loss:.4f}"
    )

    print(
        f"Train Acc  : {train_acc:.4f}"
    )

    print(
        f"Val Loss   : {val_loss:.4f}"
    )

    print(
        f"Val Acc    : {val_acc:.4f}"
    )

    print(
        f"Val AUC    : {val_auc:.4f}"
    )

    print(
        f"LR         : {current_lr:.7f}"
    )


    # --------------------------------------------------------
    # SAVE BEST MODEL USING VALIDATION LOSS
    # --------------------------------------------------------

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_state = copy.deepcopy(
            model.state_dict()
        )


        torch.save(
            {
                "model_state_dict":
                    best_state,

                "best_val_loss":
                    best_val_loss,

                "epoch":
                    epoch,

                "seed":
                    SEED
            },
            CHECKPOINT_PATH
        )


        patience_counter = 0


        print(
            "✓ Best model saved."
        )


    else:

        patience_counter += 1


        print(
            f"No improvement "
            f"({patience_counter}/"
            f"{EARLY_STOPPING_PATIENCE})"
        )


    if patience_counter >= EARLY_STOPPING_PATIENCE:

        print(
            "\nEarly stopping triggered."
        )

        break


# ============================================================
# 30. LOAD BEST CHECKPOINT
# ============================================================

print("\n======================================")
print("LOADING BEST CHECKPOINT")
print("======================================")


if not os.path.exists(
    CHECKPOINT_PATH
):

    raise FileNotFoundError(
        "Best checkpoint was not created."
    )


# IMPORTANT:
#
# Earlier you got a PyTorch 2.6+
# weights_only error.
#
# weights_only=False fixes that for a
# checkpoint created by this trusted code.

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location=DEVICE,
    weights_only=False
)


model.load_state_dict(
    checkpoint["model_state_dict"]
)

model = model.to(DEVICE)

model.eval()


print(
    "Best checkpoint loaded successfully."
)

print(
    "Best validation loss:",
    checkpoint["best_val_loss"]
)

print(
    "Best epoch:",
    checkpoint["epoch"]
)


# ============================================================
# 31. FINAL VALIDATION RESULTS
# ============================================================

val_metrics = evaluate(
    model,
    val_loader,
    criterion,
    DEVICE
)


print("\n======================================")
print("VALIDATION RESULTS")
print("======================================")


print(
    f"Accuracy  : "
    f"{val_metrics['accuracy']:.4f}"
)

print(
    f"Precision : "
    f"{val_metrics['precision']:.4f}"
)

print(
    f"Recall    : "
    f"{val_metrics['recall']:.4f}"
)

print(
    f"F1 Score  : "
    f"{val_metrics['f1']:.4f}"
)

print(
    f"AUC       : "
    f"{val_metrics['auc']:.4f}"
)


# ============================================================
# 32. FINAL TEST RESULTS
# ============================================================

test_metrics = evaluate(
    model,
    test_loader,
    criterion,
    DEVICE
)


print("\n======================================")
print("FINAL TEST RESULTS")
print("======================================")


print(
    f"Accuracy  : "
    f"{test_metrics['accuracy']:.4f}"
)

print(
    f"Precision : "
    f"{test_metrics['precision']:.4f}"
)

print(
    f"Recall    : "
    f"{test_metrics['recall']:.4f}"
)

print(
    f"F1 Score  : "
    f"{test_metrics['f1']:.4f}"
)

print(
    f"AUC       : "
    f"{test_metrics['auc']:.4f}"
)


# ============================================================
# 33. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    test_metrics["labels"],
    test_metrics["preds"]
)


print("\nConfusion Matrix:")
print(cm)


print("\nClassification Report:")

print(
    classification_report(
        test_metrics["labels"],
        test_metrics["preds"],
        target_names=[
            "Normal",
            "Glaucoma"
        ],
        zero_division=0
    )
)


# ============================================================
# 34. SAVE TEST RESULTS
# ============================================================

test_results_df = test_df.copy()

test_results_df["predicted_label"] = (
    test_metrics["preds"]
)

test_results_df["glaucoma_probability"] = (
    test_metrics["probs"]
)


test_results_df["prediction"] = (
    test_results_df["predicted_label"]
    .map({
        0: "normal",
        1: "glaucoma"
    })
)


test_results_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "combined_test_predictions.csv"
    ),
    index=False
)


# ============================================================
# 35. SAVE TRAINING HISTORY
# ============================================================

history_df = pd.DataFrame(
    history
)

history_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "training_history.csv"
    ),
    index=False
)


# ============================================================
# 36. PLOT TRAINING CURVES
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history["train_loss"],
    label="Train Loss"
)

plt.plot(
    history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(True)

plt.show()


plt.figure(figsize=(8, 5))

plt.plot(
    history["train_acc"],
    label="Train Accuracy"
)

plt.plot(
    history["val_acc"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(True)

plt.show()


# ============================================================
# 37. GRAD-CAM
# ============================================================

class GradCAM:

    def __init__(
        self,
        model,
        target_layer
    ):

        self.model = model

        self.target_layer = target_layer

        self.activations = None

        self.gradients = None


        self.forward_handle = (
            target_layer.register_forward_hook(
                self.save_activation
            )
        )

        self.backward_handle = (
            target_layer.register_full_backward_hook(
                self.save_gradient
            )
        )


    def save_activation(
        self,
        module,
        input,
        output
    ):

        self.activations = output


    def save_gradient(
        self,
        module,
        grad_input,
        grad_output
    ):

        self.gradients = grad_output[0]


    def generate(
        self,
        image_tensor,
        target_class=None
    ):

        self.model.eval()

        image_tensor = image_tensor.to(
            DEVICE
        )

        self.model.zero_grad()


        logits = self.model(
            image_tensor
        )


        if target_class is None:

            target_class = torch.argmax(
                logits,
                dim=1
            ).item()


        score = logits[
            0,
            target_class
        ]


        score.backward()


        activations = (
            self.activations
        )

        gradients = (
            self.gradients
        )


        weights = gradients.mean(
            dim=(2, 3),
            keepdim=True
        )


        cam = (
            weights *
            activations
        ).sum(
            dim=1
        )


        cam = torch.relu(
            cam
        )


        cam = cam.squeeze().detach().cpu().numpy()


        cam -= cam.min()


        if cam.max() > 0:

            cam /= cam.max()


        return cam, target_class


    def close(self):

        self.forward_handle.remove()

        self.backward_handle.remove()


# ============================================================
# 38. GRAD-CAM TARGET LAYER
# ============================================================

# Last ConvNeXt feature stage

target_layer = model.backbone.features[7]

gradcam = GradCAM(
    model,
    target_layer
)


print(
    "\nGrad-CAM initialized successfully."
)


# ============================================================
# 39. GENERATE SAMPLE GRAD-CAM
# ============================================================

def denormalize_image(
    tensor
):

    image = tensor.cpu().numpy()

    image = np.transpose(
        image,
        (1, 2, 0)
    )

    mean = np.array([
        0.485,
        0.456,
        0.406
    ])

    std = np.array([
        0.229,
        0.224,
        0.225
    ])

    image = (
        image * std
    ) + mean

    image = np.clip(
        image,
        0,
        1
    )

    return image


sample_image, sample_label = (
    test_dataset[0]
)


cam, predicted_class = gradcam.generate(
    sample_image.unsqueeze(0),
    target_class=None
)


original_image = denormalize_image(
    sample_image
)


plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)

plt.imshow(
    original_image
)

plt.title(
    f"Original\nTrue: "
    f"{'Glaucoma' if sample_label == 1 else 'Normal'}"
)

plt.axis("off")


plt.subplot(1, 2, 2)

plt.imshow(
    original_image
)

plt.imshow(
    cam,
    cmap="jet",
    alpha=0.45
)

plt.title(
    f"Grad-CAM\nPredicted: "
    f"{'Glaucoma' if predicted_class == 1 else 'Normal'}"
)

plt.axis("off")

plt.tight_layout()

plt.show()


# ============================================================
# CORRECTED VLM PART
# Pretrained BLIP + Grad-CAM Highlighted Image
# NO FINE-TUNING
# ============================================================

from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import torch
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD PRETRAINED BLIP
# ============================================================

VLM_MODEL_NAME = "Salesforce/blip-image-captioning-base"

print("Loading pretrained VLM...")

processor = BlipProcessor.from_pretrained(
    VLM_MODEL_NAME
)

vlm_model = BlipForConditionalGeneration.from_pretrained(
    VLM_MODEL_NAME
)

vlm_model = vlm_model.to(DEVICE)

vlm_model.eval()

print("VLM loaded successfully.")
print("Fine-tuning: NO")


# ============================================================
# 2. CREATE GRAD-CAM OVERLAY IMAGE
# ============================================================

def create_gradcam_overlay(
    image_tensor,
    cam
):

    # Convert normalized tensor back to image
    image = image_tensor.detach().cpu().numpy()

    image = np.transpose(
        image,
        (1, 2, 0)
    )

    mean = np.array([
        0.485,
        0.456,
        0.406
    ])

    std = np.array([
        0.229,
        0.224,
        0.225
    ])

    image = (
        image * std
    ) + mean

    image = np.clip(
        image,
        0,
        1
    )

    # Convert CAM to 0-255
    cam_uint8 = np.uint8(
        255 * cam
    )

    # Resize CAM
    cam_img = Image.fromarray(
        cam_uint8
    ).resize(
        (image.shape[1], image.shape[0])
    )

    cam_resized = np.array(
        cam_img
    ) / 255.0

    # Create jet heatmap
    heatmap = plt.get_cmap("jet")(
        cam_resized
    )[:, :, :3]

    # Overlay
    overlay = (
        0.55 * image +
        0.45 * heatmap
    )

    overlay = np.clip(
        overlay,
        0,
        1
    )

    overlay_uint8 = np.uint8(
        overlay * 255
    )

    return Image.fromarray(
        overlay_uint8
    )


# ============================================================
# 3. VLM EXPLANATION FUNCTION
# ============================================================

def generate_vlm_explanation(
    image,
    predicted_class,
    probability,
    cam=None
):

    # --------------------------------------------------------
    # If Grad-CAM is available, give VLM the highlighted image
    # --------------------------------------------------------

    if cam is not None:

        image_for_vlm = create_gradcam_overlay(
            image,
            cam
        )

    else:

        # Convert tensor to PIL image
        image_np = image.detach().cpu().numpy()

        image_np = np.transpose(
            image_np,
            (1, 2, 0)
        )

        mean = np.array([
            0.485,
            0.456,
            0.406
        ])

        std = np.array([
            0.229,
            0.224,
            0.225
        ])

        image_np = (
            image_np * std
        ) + mean

        image_np = np.clip(
            image_np,
            0,
            1
        )

        image_for_vlm = Image.fromarray(
            np.uint8(image_np * 255)
        )


    # --------------------------------------------------------
    # Prediction text
    # --------------------------------------------------------

    if predicted_class == 1:

        prediction_text = (
            "The computer vision model predicted "
            "GLAUCOMA."
        )

    else:

        prediction_text = (
            "The computer vision model predicted "
            "NORMAL."
        )


    # --------------------------------------------------------
    # VLM prompt
    # --------------------------------------------------------
    #
    # We tell the VLM:
    # - image is retinal fundus
    # - Grad-CAM highlighted the model-attended region
    # - describe visible evidence
    # - don't invent measurements
    #
    # --------------------------------------------------------

    prompt = f"""
A retinal fundus photograph is shown with a heatmap
highlighting the region used by an image classification model.

{prediction_text}

Describe the visible retinal features in the highlighted
region that may be relevant to the classification.

Focus on:
- optic disc appearance
- optic cup appearance if clearly visible
- retinal structure
- visible suspicious or non-suspicious characteristics

Do not invent numerical measurements.
Do not claim a medical diagnosis from the image alone.
Describe only visually observable characteristics.
"""


    # --------------------------------------------------------
    # Process image + prompt
    # --------------------------------------------------------

    inputs = processor(
        images=image_for_vlm,
        text=prompt,
        return_tensors="pt"
    )


    inputs = {
        key: value.to(DEVICE)
        for key, value in inputs.items()
    }


    # --------------------------------------------------------
    # Generate explanation
    # --------------------------------------------------------

    with torch.no_grad():

        output = vlm_model.generate(
            **inputs,
            max_new_tokens=100,
            num_beams=5,
            repetition_penalty=1.3,
            no_repeat_ngram_size=3
        )


    explanation = processor.decode(
        output[0],
        skip_special_tokens=True
    )


    # --------------------------------------------------------
    # Remove prompt from output if BLIP repeats it
    # --------------------------------------------------------

    explanation = explanation.strip()


    return explanation


# ============================================================
# 4. TEST VLM WITH YOUR EXISTING TEST IMAGE
# ============================================================

sample_image, sample_label = test_dataset[0]

sample_path = test_df.iloc[0]["image_path"]


# ------------------------------------------------------------
# Get model prediction
# ------------------------------------------------------------

model.eval()

with torch.no_grad():

    input_tensor = sample_image.unsqueeze(
        0
    ).to(DEVICE)

    logits = model(
        input_tensor
    )

    probabilities = torch.softmax(
        logits,
        dim=1
    )

    predicted_class = torch.argmax(
        probabilities,
        dim=1
    ).item()

    glaucoma_probability = probabilities[
        0, 1
    ].item()


# ============================================================
# 5. GENERATE GRAD-CAM
# ============================================================

cam, predicted_class = gradcam.generate(
    sample_image.unsqueeze(0),
    target_class=predicted_class
)


# ============================================================
# 6. GENERATE VLM EXPLANATION
# ============================================================

vlm_explanation = generate_vlm_explanation(
    image=sample_image,
    predicted_class=predicted_class,
    probability=glaucoma_probability,
    cam=cam
)


# ============================================================
# 7. DISPLAY FINAL RESULT
# ============================================================

print("\n")
print("=" * 60)
print("MODEL + GRAD-CAM + VLM RESULT")
print("=" * 60)

print(
    "\nTrue Class:",
    "GLAUCOMA" if sample_label == 1 else "NORMAL"
)

print(
    "Predicted Class:",
    "GLAUCOMA" if predicted_class == 1 else "NORMAL"
)

print(
    f"Glaucoma Probability: "
    f"{glaucoma_probability:.4f}"
)

print("\nVLM Visual Explanation:")
print(vlm_explanation)


# ============================================================
# 8. DISPLAY IMAGE + GRAD-CAM
# ============================================================

overlay_image = create_gradcam_overlay(
    sample_image,
    cam
)


plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)

plt.imshow(
    denormalize_image(sample_image)
)

plt.title(
    "Original Fundus Image"
)

plt.axis("off")


plt.subplot(1, 2, 2)

plt.imshow(
    overlay_image
)

plt.title(
    "Grad-CAM Highlighted Image"
)

plt.axis("off")


plt.tight_layout()

plt.show()




# ============================================================
# 43. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("EXPERIMENT COMPLETED")
print("=" * 60)

print(
    f"ACRIMA images     : {len(acrima_df)}"
)

print(
    f"RIM-ONE images    : {len(rim_df)}"
)

print(
    f"Combined images   : {len(combined_df)}"
)

print(
    f"Training images   : {len(train_df)}"
)

print(
    f"Validation images : {len(val_df)}"
)

print(
    f"Testing images    : {len(test_df)}"
)

print("\nModel:")
print(
    "ConvNeXt-Tiny + Multi-Level Feature Fusion"
)

print("\nAugmentation:")
print(
    "YES - Training only"
)

print(
    "NO - Validation/Test"
)

print("\nExplainability:")
print(
    "Grad-CAM"
)

print("\nVLM:")
print(
    "BLIP image-captioning-base"
)

print(
    "Fine-tuning: NO"
)

print("\nBest checkpoint:")
print(
    CHECKPOINT_PATH
)

print("\nResults:")
print(
    RESULTS_DIR
)

print("\nDone.")
```
